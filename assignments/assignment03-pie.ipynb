```python
# Import modules we will need
import pandas as pd
import seaborn as sb
import matplotlib.pyplot as plt
```

```python
# define the links to the URL
url = "https://drive.google.com/uc?id=1AWPf-pJodJKeHsARQK_RHiNsE8fjPCVK&export=download"
data = pd.read_csv(url) 

#select only the email address column
emails = data.iloc[:,5]
#print it out
print(emails)
```

```python
# Convert emails into a list of strings so we may parse them by the @ character
emails = emails.astype(str)

#strip out the text before the domains using the partition method
domains = emails.str.partition('@')[2]
unique_domains = domains.unique()

#Count each individual domain
count_org = domains.value_counts().get('example.org')
print("Count of example.org:",count_org)
count_net = domains.value_counts().get('example.net')
print("Count of example.net:",count_net)
count_com = domains.value_counts().get('example.com')
print("Count of example.com:",count_com)
```

```python
# Create piechart
# https://www.geeksforgeeks.org/python/how-to-create-a-pie-chart-in-seaborn/
# combine the counts into one series. 
domaindata = [count_org,count_net,count_com]
domainlabels = ['example.org','example.net','example.com']
colors = sb.color_palette('pastel')
title = ("Piechart of Email Domains")

fig, ax = plt.subplots()
wedges, texts, autotexts = ax.pie(
    domaindata,
    labels=domainlabels,
    colors=colors,
    startangle=90,
    autopct='%.1f%%',
    wedgeprops=dict(width=0.6)
)

ax.legend( domainlabels, title="Domains", loc="center left", bbox_to_anchor=(1, 0, 0.5, 1))
plt.show()
```
